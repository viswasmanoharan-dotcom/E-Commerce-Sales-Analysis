<h1 style="text-align:center; color:red;"> Column Description</h1>

<hr>

<p style="text-align:center;">

</p>

- **Order_ID:** Unique identifier assigned to each customer order.
- **Customer_ID:** Unique identifier representing the customer who placed the order.
- **Order_Date:** Date on which the order was placed.
- **Product_Category:** Category or classification of the product purchased.
- **Product_Name:** Name of the product included in the order.
- **Quantity:** Number of units of the product purchased in the order.
- **Unit_Price_USD:** Price of one unit of the product in US dollars.
- **Discount_Percent:** Percentage discount applied to the product or order.
- **Payment_Method:** Payment method used by the customer to complete the transaction.
- **Shipping_City:** City to which the order was shipped.
- **Country:** Country associated with the order or shipping destination.
- **Order_Status:** Current status of the order, such as completed, pending, cancelled, or returned.
- **Customer_Rating:** Rating provided by the customer for the purchase/order experience.

 # Handling Outliers

```python
Q1 = df['Unit_Price_USD'].quantile(0.25)
Q3 = df['Unit_Price_USD'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['Unit_Price_USD'] < lower) |
    (df['Unit_Price_USD'] > upper)
]

print(outliers[['Product_Category', 'Product_Name', 'Unit_Price_USD']].head(20))
```

```python
Q1 = df['Unit_Price_USD'].quantile(0.25)
Q3 = df['Unit_Price_USD'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_mask = (
    (df['Unit_Price_USD'] < lower_bound) |
    (df['Unit_Price_USD'] > upper_bound)
)

print("Outliers before removal:", outlier_mask.sum())

df = df[~outlier_mask].copy()

print("Outliers after removal:", (
    (df['Unit_Price_USD'] < lower_bound) |
    (df['Unit_Price_USD'] > upper_bound)
).sum())
```

```python
Q1 = df['Unit_Price_USD'].quantile(0.25)
Q3 = df['Unit_Price_USD'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_mask = (
    (df['Unit_Price_USD'] < lower_bound) |
    (df['Unit_Price_USD'] > upper_bound)
)

# Store outliers before removal
outliers_df = df[outlier_mask].copy()

print("Outliers before removal:", len(outliers_df))

# Save outliers
outliers_df.to_csv("Unit_Price_USD_outliers.csv", index=False)
outliers_df.to_excel("Unit_Price_USD_outliers.xlsx", index=False)

# Remove outliers
df = df[~outlier_mask].copy()

print("Rows after outlier removal:", len(df))
```

## Revenue Contribution by Category

```python
category_percentage = (
    df.groupby('Product_Category')['Net_Sales']
      .sum()
      .div(df['Net_Sales'].sum())
      .mul(100)
      .round(2)
      .reset_index(name='Sales_Percentage'))

category_percentage
```

## Average Order Value By Category (AOV)

```python
aov_category = (
    df.groupby('Product_Category')
      .apply(lambda x: x.groupby('Order_ID')['Net_Sales'].sum().mean())
      .sort_values(ascending=False)
)

aov_category
```
