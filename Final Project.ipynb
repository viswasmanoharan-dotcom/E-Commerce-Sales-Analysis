## Step 2: Data Pre-Processing

```python
# Handle missing values

df['Discount_Percent'] = df['Discount_Percent'].fillna(0)

df['Shipping_City'] = df['Shipping_City'].fillna('Unknown')

df['Customer_Rating'] = df['Customer_Rating'].fillna(df['Customer_Rating'].median())

# Check missing values again
df.isnull().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
# Step 1: Parse the mixed date formats into datetime objects
df["Order_Date"] = pd.to_datetime(df["Order_Date"], format="mixed")

# Step 2: Format the dates into dd-mm-yy (e.g., '31-12-25')
df["Order_Date"] = df["Order_Date"].dt.strftime("%d-%m-%y")
```

 ## Step 3:Handling Missing Data

```python
#  Handle Missing Text Values in Shipping_City
df["Shipping_City"] = df["Shipping_City"].fillna("Unknown")
```

```python
# View first 10 converted values
print(df["Shipping_City"].head(30))

# View last 10 converted values
print(df["Shipping_City"].tail(10))
```

```python
#  Command to calculate and add the new column
df["Gross_Revenue_USD"] = df["Quantity"] * df["Unit_Price_USD"]

#  Commands to view the new column

# View the first 5 rows containing the relevant columns
print(df[["Order_ID", "Quantity", "Unit_Price_USD", "Gross_Revenue_USD"]].head())

# View summary statistics for Gross_Revenue_USD
print(df["Gross_Revenue_USD"].describe())
```

```python
# 3. Fill missing Discount_Percent values with 0 and calculate Discount_Amount_USD
df["Discount_Amount_USD"] = (
    df["Gross_Revenue_USD"] * (df["Discount_Percent"].fillna(0) / 100)
).round(2)

#  View the results
print(
    df[
        [
            "Order_ID",
            "Gross_Revenue_USD",
            "Discount_Percent",
            "Discount_Amount_USD",
        ]
    ].head(10)
)
```

## Adding new Columns

```python
#  Calculate Net Revenue
df["Net_Revenue_USD"] = (
    df["Gross_Revenue_USD"] - df["Discount_Amount_USD"]
).round(2)

# View the result
print(
    df[
        [
            "Order_ID",
            "Gross_Revenue_USD",
            "Discount_Amount_USD",
            "Net_Revenue_USD",
        ]
    ].head(10)
)
```

 # Handling Outliers

```python
Q1 = df['Unit_Price_USD'].quantile(0.25)
Q3 = df['Unit_Price_USD'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['Unit_Price_USD'] < lower) |
    (df['Unit_Price_USD'] > upper)
]

print(outliers[['Product_Category', 'Product_Name', 'Unit_Price_USD']].head(20))
```

```python
Q1 = df['Unit_Price_USD'].quantile(0.25)
Q3 = df['Unit_Price_USD'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_mask = (
    (df['Unit_Price_USD'] < lower_bound) |
    (df['Unit_Price_USD'] > upper_bound)
)

print("Outliers before removal:", outlier_mask.sum())

df = df[~outlier_mask].copy()

print("Outliers after removal:", (
    (df['Unit_Price_USD'] < lower_bound) |
    (df['Unit_Price_USD'] > upper_bound)
).sum())
```
